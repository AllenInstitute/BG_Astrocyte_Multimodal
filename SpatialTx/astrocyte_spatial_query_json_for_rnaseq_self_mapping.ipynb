```python
import json
import anndata as ad
```

# Load rna-seq & spatial data

```python
adata_rnaseq = ad.read_h5ad(
    '/data/snRNAseq_hmba_bg_human_astro_3subGroups/HMBA_human_Astro_3subGroups.h5ad',
    backed='r'
)
adata_rnaseq
```

```python
adata_spatial = ad.read_h5ad(
    '/data/cirro_files_hmba_human_20250922/hmba_release_format_human_cirro_h5ad_20250922.h5ad',
    backed='r'
)
adata_spatial
```

# Get gene sets

```python
human_spatial_gene_panel = sorted([g for g in adata_spatial.var_names if 'Blank' not in g])

display(len(human_spatial_gene_panel))
```

```python
human_rnaseq_genes = sorted(adata_rnaseq.var_names.tolist())
display(len(human_rnaseq_genes))
```

```python
# Check if spatial genes are in rnaseq gene list
gene_set_verified = [g for g in human_spatial_gene_panel if g in human_rnaseq_genes]
display(len(gene_set_verified))


if len(gene_set_verified) != len(human_spatial_gene_panel):
    display('missing genes:', set(human_spatial_gene_panel) - set(human_rnaseq_genes))
        
```

```python
gene_set_name = 'human_MERSCOPE_BG_gene_panel'

# make a flat mappting query marker file
query_markers_dict = {
    'None': gene_set_verified
}
print(f"{len(gene_set_verified)} genes: {gene_set_verified}")

# write out to json
with open(f"/results/query_markers_dispatch_{gene_set_name}.json", "w") as f:
    json.dump(query_markers_dict, f)
```
